### What is the relationship between a projectile and an ellipse?

- Projectile motion (flat space): Constant downward gravity (
g
g), trajectory is a parabola.

- Newton's Cannonball: The classic thought experiment that bridges the two. Throw a ball fast enough, and the Earth curves away beneath it as fast as it falls. The trajectory becomes an ellipse (orbit) instead of a parabola.

- They are both governed by the same force (Gravity) and the same mathematical family (Conic Sections). A projectile is just a "partial ellipse" or a "very stretched ellipse" cut off by the ground.

###  A function that simulate the fall of an apple or the launch of a rocket, use this function to find out where it will land after (R)
where :
- s0 : the start position
- vh: hirozontal vilocity
- R: the rang, where the object falls after

```python
import math

def calculate_range(s0, vh, g=9.81):

    #calaculate the time to reach the ground
    time_to_fall = math.sqrt((2 * s0) / g)
    
    # calc the hirozontal range
    R = vh * time_to_fall
    return R

print(calculate_range(20, 5)) 
```

### Drawing an Ellipse (Geometry for Eclipse Simulation):
#### To draw an elliptical orbit (like the Moon around Earth or Earth around the Sun), we define:

- t (Parameter/Angle): The changing variable (from 0 to 2π) that determines the position of the "pen" at any given moment. As changes,the pen draws the curve.


- a (Semi-major axis): The longest radius of the ellipse (horizontal stretch).


- b (Semi-minor axis): The shortest radius of the ellipse (vertical stretch).



```python
import numpy as np
import matplotlib.pyplot as plt

t= np.linspace(0, 2*np.pi, 100)
a= 5
b= 3
x= a * np.cos(t)
y= b * np.sin(t) 


plt.plot(x,y)
plt.axis("equal")
plt.show()
```

```python

a = 10  #a (Semi-major axis)
e = 0.5 #Eccentricity 

# Angles (theta) from 0 to 360 degrees (in radians)
theta = np.linspace(0, 2*np.pi, 100)

# Kepler's polar equation for calculating the distance r
r = (a * (1 - e**2)) / (1 + e * np.cos(theta))

# Convert polar coordinates to Cartesian coordinates for plotting
x = r * np.cos(theta)
y = r * np.sin(theta)

plt.plot(x, y)
plt.plot(0, 0, 'yo', markersize=10, label='Sun (Focus)') # The Sun at F
plt.axis('equal')
plt.legend()
plt.show()
```
